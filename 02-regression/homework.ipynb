# Q3

```python
def train_linear_regression(X, y):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    XTX = X.T.dot(X)
    XTX_inv = np.linalg.inv(XTX)
    w_full = XTX_inv.dot(X.T).dot(y)

    return w_full[0], w_full[1:]
```

```python
def rmse(y, y_pred):
    error = y - y_pred
    se = error ** 2
    mse = se.mean()
    return np.sqrt(mse)
```

## With 0

```python
n = len(df)
n_val = int(n * 0.2)
n_test = int(n * 0.2)
n_train = n - n_val - n_test

np.random.seed(42)
idx = np.arange(n)
np.random.shuffle(idx)

df_train = df.iloc[idx[:n_train]]
df_val = df.iloc[idx[n_train:n_train + n_val]]
df_test = df.iloc[idx[n_train + n_val:]]
```

```python
df_test.horsepower = df_test.horsepower.fillna(0)
df_val.horsepower = df_val.horsepower.fillna(0)
df_train.horsepower = df_train.horsepower.fillna(0)
```

```python
X_train = df_train[["engine_displacement", "horsepower", "vehicle_weight", "model_year"]].values
X_train
```

```python
y_train = df_train["fuel_efficiency_mpg"].values
y_train
```

```python
w0, w = train_linear_regression(X_train, y_train)
```

```python
y_pred = w0 + X_train.dot(w)
y_pred
```

```python
X_val = df_val[["engine_displacement", "horsepower", "vehicle_weight", "model_year"]].values
X_val
```

```python
y_val = df_val["fuel_efficiency_mpg"].values
y_val
```

```python
y_pred = w0 + X_val.dot(w)
y_pred
```

```python
rmse_with_0 = round(rmse(y_val, y_pred), 3)
rmse_with_0
```
